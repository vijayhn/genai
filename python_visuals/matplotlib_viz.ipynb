```python
from matplotlib.pyplot import figure, show
import matplotlib.pyplot as plt
import pandas as pd
```

```python
coffee_sales=pd.read_csv("Coffe_sales.csv")
big_mart_sales=pd.read_csv("big_mart_sales.csv")
```

```python
coffee_sales.head()
```

```python
big_mart_sales.head()
```

```python
# coffee_sales['money'].groupby(coffee_sales['hour_of_day']).sum().reset_index()['money']
# coffee_sales['hour_of_day'].value_counts().sort_index()
ax=coffee_sales.groupby('hour_of_day')['money'].sum().reset_index()\
    .plot(kind='bar', x='hour_of_day', y='money', title='Coffee Sales by Hour of Day'\
          ,label='Money', color='brown',figsize=(10,5), legend=True)
ax.bar_label(ax.containers[0], fmt='%.1f', padding=3)
plt.xticks(rotation=90)
plt.show()
```

```python
big_mart_sales.groupby('Outlet_Location_Type').agg({'Item_Outlet_Sales': 'sum', 'Item_MRP': 'mean'}).reset_index()\
    .plot(kind='bar', x='Outlet_Location_Type'
           ,secondary_y='Item_MRP', title='Mart Sales vs. MRP')
```

```python
ax=big_mart_sales.groupby('Outlet_Location_Type').agg({'Item_Outlet_Sales': 'sum', 'Item_MRP': 'mean'}).reset_index()\
    .plot(kind='bar', x='Outlet_Location_Type',secondary_y='Item_MRP', title='Mart Sales vs. MRP'\
          ,label=['Total Sales','Avg. MRP'], color=['brown','green'],figsize=(10,5), legend=True)
ax.bar_label(ax.containers[0], fmt='$.1f', padding=3)
for container in ax.right_ax.containers:
    ax.right_ax.bar_label(container, fmt='%.1f', padding=3)
plt.xticks(rotation=90)
plt.show()
```
